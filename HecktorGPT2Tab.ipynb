```python

# ============================================================
# HECKTOR — GPT-2 TABULAR CLASSIFIER
# Protocol aligned with Hancock GPT-2 Tabular
# Seed 42 | 65/15/20 split | Batch 64 | 500 epochs
# One-hot encoding | No SMOTE | No hard/gradient clipping
# Four validation-selected checkpoints
# ============================================================

!pip -q install gdown openpyxl

import os
import json
import random
import copy
import warnings
from pathlib import Path

import gdown
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import joblib

from torch.utils.data import TensorDataset, DataLoader
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.metrics import (
    balanced_accuracy_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    precision_recall_curve
)
from transformers import GPT2Model

warnings.filterwarnings("ignore", category=FutureWarning)

# ============================================================
# 1. SETTINGS AND REPRODUCIBILITY
# ============================================================

SEED = 42
MAX_EPOCHS = 500
BATCH_SIZE = 64
LEARNING_RATE = 1e-3
WEIGHT_DECAY = 1e-4
DROPOUT = 0.1
MODEL_NAME = "openai-community/gpt2"

def seed_everything(seed=42):
    os.environ["PYTHONHASHSEED"] = str(seed)
    os.environ["CUBLAS_WORKSPACE_CONFIG"] = ":4096:8"

    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    torch.use_deterministic_algorithms(True, warn_only=True)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
        torch.backends.cuda.matmul.allow_tf32 = False
        torch.backends.cudnn.allow_tf32 = False

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

seed_everything(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

OUT_DIR = Path("HECKTOR_GPT2_Tabular_Hancock_Aligned")
CHECKPOINT_DIR = OUT_DIR / "checkpoints"
OUT_DIR.mkdir(parents=True, exist_ok=True)
CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)

# ============================================================
# 2. LOAD HECKTOR DATA
# ============================================================

DATA_FILE = "HPV2025.xlsx"
FILE_ID = "1DGDH2e6dEkpdwwtCvH6ds1CwJT2Hdmkv"

if not os.path.exists(DATA_FILE):
    downloaded = gdown.download(
        id=FILE_ID,
        output=DATA_FILE,
        quiet=False
    )
    if downloaded is None or not os.path.exists(DATA_FILE):
        raise FileNotFoundError("Could not download HPV2025.xlsx.")

df = pd.read_excel(DATA_FILE)

TARGET = "HPV Status"

# Keep the predictors used by the recent clinical narrative model.
FEATURES = [
    "Age",
    "Gender",
    "Tobacco Consumption",
    "Alcohol Consumption",
    "Performance Status",
    "T-stage",
    "N-stage",
    "M-stage"
]

missing = [c for c in [TARGET] + FEATURES if c not in df.columns]
if missing:
    raise KeyError(f"Required Excel columns missing: {missing}")

# Explicitly exclude identifiers, task columns, and removed predictors.
columns_to_remove = [
    "PatientID", "CenterID",
    "Task 1", "Task 2", "Task 3",
    "Relapse", "RFS", "Treatment"
]
df = df.drop(
    columns=[c for c in columns_to_remove if c in df.columns],
    errors="ignore"
)

# Do not convert T/N/M stages into continuous numbers.
# They will be one-hot encoded as categorical predictors.

# ============================================================
# 3. NORMALISE TARGET
# ============================================================

def normalize_target(value):
    if pd.isna(value):
        return np.nan

    s = str(value).strip().lower()

    if s in {"0", "0.0", "negative", "neg", "false", "no"}:
        return 0
    if s in {"1", "1.0", "positive", "pos", "true", "yes"}:
        return 1

    try:
        n = float(s)
        if n in (0, 1):
            return int(n)
    except (ValueError, TypeError):
        pass

    return np.nan

df["HPV_Label"] = df[TARGET].map(normalize_target)

invalid_target_rows = int(df["HPV_Label"].isna().sum())
df = df.dropna(subset=["HPV_Label"]).copy()
df["HPV_Label"] = df["HPV_Label"].astype(int)

df = df[FEATURES + ["HPV_Label"]].copy()

if df.empty:
    raise ValueError("No usable HPV labels found.")

class_counts_all = df["HPV_Label"].value_counts().reindex(
    [0, 1], fill_value=0
)

if (class_counts_all < 2).any():
    raise ValueError(
        f"Insufficient examples in one HPV class: "
        f"{class_counts_all.to_dict()}"
    )

print("Dataset shape:", df.shape)
print("Rows removed for missing/invalid target:", invalid_target_rows)
print("Class counts (0=negative, 1=positive):")
print(class_counts_all)

# ============================================================
# 4. STRATIFIED 65/15/20 SPLIT
# ============================================================

train_val_df, test_df = train_test_split(
    df,
    test_size=0.20,
    random_state=SEED,
    stratify=df["HPV_Label"]
)

train_df, val_df = train_test_split(
    train_val_df,
    test_size=0.1875,
    random_state=SEED,
    stratify=train_val_df["HPV_Label"]
)

train_df, val_df, test_df = [
    part.reset_index(drop=True)
    for part in (train_df, val_df, test_df)
]

for name, part in [
    ("Train", train_df),
    ("Validation", val_df),
    ("Test", test_df)
]:
    print(
        f"{name}: n={len(part)}, class counts="
        f"{part['HPV_Label'].value_counts().reindex([0,1], fill_value=0).to_dict()}"
    )

X_train_df = train_df[FEATURES].copy()
X_val_df = val_df[FEATURES].copy()
X_test_df = test_df[FEATURES].copy()

y_train = train_df["HPV_Label"].to_numpy(dtype=np.int64)
y_val = val_df["HPV_Label"].to_numpy(dtype=np.int64)
y_test = test_df["HPV_Label"].to_numpy(dtype=np.int64)

# ============================================================
# 5. TRAIN-ONLY IMPUTATION AND ONE-HOT ENCODING
# ============================================================

# Match Hancock's tabular setup: Age is numeric;
# all remaining predictors are categorical.
NUMERIC_FEATURES = ["Age"]
CATEGORICAL_FEATURES = [c for c in FEATURES if c not in NUMERIC_FEATURES]

for frame in (X_train_df, X_val_df, X_test_df):
    frame["Age"] = pd.to_numeric(frame["Age"], errors="coerce")

age_median = X_train_df["Age"].median()
if pd.isna(age_median):
    age_median = 0.0

for frame in (X_train_df, X_val_df, X_test_df):
    frame["Age"] = frame["Age"].fillna(age_median)

# Treat categorical predictors as categories, including stages.
for col in CATEGORICAL_FEATURES:
    X_train_df[col] = X_train_df[col].where(
        X_train_df[col].notna(), np.nan
    )
    modes = X_train_df[col].dropna().astype(str).mode()
    fill_value = modes.iloc[0] if not modes.empty else "Unknown"

    for frame in (X_train_df, X_val_df, X_test_df):
        frame[col] = frame[col].fillna(fill_value).astype(str)
        frame[col] = frame[col].replace({
            "": "Unknown",
            "nan": "Unknown",
            "None": "Unknown"
        })

try:
    onehot = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )
except TypeError:
    onehot = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )

preprocessor = ColumnTransformer([
    ("numeric", StandardScaler(), NUMERIC_FEATURES),
    ("categorical", onehot, CATEGORICAL_FEATURES)
])

# Fit preprocessing on training data only.
X_train = preprocessor.fit_transform(X_train_df).astype(np.float32)
X_val = preprocessor.transform(X_val_df).astype(np.float32)
X_test = preprocessor.transform(X_test_df).astype(np.float32)

print("Predictors:", FEATURES)
print("Encoded input dimension:", X_train.shape[1])

# ============================================================
# 6. CLASS WEIGHTS AND DATA LOADER — NO SMOTE
# ============================================================

counts = np.bincount(y_train, minlength=2)

if np.any(counts == 0):
    raise ValueError("Both classes must occur in training data.")

weights_np = len(y_train) / (2.0 * counts)

class_weights = torch.tensor(
    weights_np,
    dtype=torch.float32,
    device=device
)

print("Training class weights:", weights_np)

train_dataset = TensorDataset(
    torch.tensor(X_train, dtype=torch.float32),
    torch.tensor(y_train, dtype=torch.long)
)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    generator=torch.Generator().manual_seed(SEED),
    num_workers=0,
    pin_memory=torch.cuda.is_available()
)

X_val_t = torch.tensor(X_val, dtype=torch.float32, device=device)
y_val_t = torch.tensor(y_val, dtype=torch.long, device=device)
X_test_t = torch.tensor(X_test, dtype=torch.float32, device=device)

# ============================================================
# 7. GPT-2 TABULAR MODEL — SAME ARCHITECTURE AS HANCOCK
# ============================================================

class GPT2TabularClassifier(nn.Module):
    def __init__(self, n_features, dropout=0.1):
        super().__init__()

        self.gpt2 = GPT2Model.from_pretrained(MODEL_NAME)
        self.gpt2.config.use_cache = False

        hidden = self.gpt2.config.hidden_size

        self.feature_projection = nn.Linear(1, hidden)

        self.feature_embedding = nn.Parameter(
            torch.zeros(1, n_features, hidden)
        )
        nn.init.normal_(
            self.feature_embedding,
            mean=0.0,
            std=0.02
        )

        self.classifier = nn.Sequential(
            nn.LayerNorm(hidden),
            nn.Dropout(dropout),
            nn.Linear(hidden, 128),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(128, 2)
        )

    def forward(self, x):
        tokens = self.feature_projection(x.unsqueeze(-1))
        tokens = tokens + self.feature_embedding

        hidden_states = self.gpt2(
            inputs_embeds=tokens,
            use_cache=False,
            return_dict=True
        ).last_hidden_state

        pooled = hidden_states.mean(dim=1)
        return self.classifier(pooled)

model = GPT2TabularClassifier(
    n_features=X_train.shape[1],
    dropout=DROPOUT
).to(device)

print(
    "Trainable parameters:",
    sum(p.numel() for p in model.parameters() if p.requires_grad)
)

# Same custom weighted loss as Hancock.
def weighted_classification_loss(logits, labels):
    losses = nn.functional.cross_entropy(
        logits,
        labels,
        reduction="none"
    )
    return (losses * class_weights[labels]).mean()

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

# ============================================================
# 8. PREDICTIONS, VALIDATION LOSS, AND THRESHOLD SELECTION
# ============================================================

def predict_probabilities(X_tensor):
    model.eval()
    with torch.no_grad():
        return torch.softmax(model(X_tensor), dim=1)[:, 1].cpu().numpy()

def evaluate_loss():
    model.eval()
    with torch.no_grad():
        return weighted_classification_loss(
            model(X_val_t), y_val_t
        ).item()

def choose_threshold(scores, labels):
    thresholds = np.unique(np.r_[0.0, scores, 1.0])
    values = [
        (
            float(t),
            balanced_accuracy_score(labels, scores >= t)
        )
        for t in thresholds
    ]
    # Maximise balanced accuracy; ties closest to 0.5.
    return max(values, key=lambda x: (x[1], -abs(x[0] - 0.5)))

# ============================================================
# 9. CHECKPOINTS SELECTED USING VALIDATION DATA ONLY
# ============================================================

checkpoint_specs = {
    "Validation_Loss": ("best_model_validation_loss.pt", "min"),
    "Validation_ROC_AUC": ("best_model_validation_roc_auc.pt", "max"),
    "Validation_PR_AUC": ("best_model_validation_pr_auc.pt", "max"),
    "Validation_Balanced_Accuracy": (
        "best_model_validation_balanced_accuracy.pt", "max"
    )
}

best_values = {
    "Validation_Loss": float("inf"),
    "Validation_ROC_AUC": -float("inf"),
    "Validation_PR_AUC": -float("inf"),
    "Validation_Balanced_Accuracy": -float("inf")
}

best_info = {}
history = []

# ============================================================
# 10. TRAIN MODEL — 500 EPOCHS, NO GRADIENT CLIPPING
# ============================================================

for epoch in range(1, MAX_EPOCHS + 1):
    model.train()
    loss_sum, n_sum = 0.0, 0

    for xb, yb in train_loader:
        xb = xb.to(device)
        yb = yb.to(device)

        optimizer.zero_grad(set_to_none=True)

        loss = weighted_classification_loss(model(xb), yb)
        loss.backward()

        # Intentionally no clip_grad_norm_ or hard clipping.
        optimizer.step()

        n = len(yb)
        loss_sum += loss.item() * n
        n_sum += n

    train_loss = loss_sum / max(n_sum, 1)
    val_loss = evaluate_loss()
    val_scores = predict_probabilities(X_val_t)

    val_auc = roc_auc_score(y_val, val_scores)
    val_ap = average_precision_score(y_val, val_scores)
    threshold, val_ba = choose_threshold(val_scores, y_val)

    current_metrics = {
        "Validation_Loss": float(val_loss),
        "Validation_ROC_AUC": float(val_auc),
        "Validation_PR_AUC": float(val_ap),
        "Validation_Balanced_Accuracy": float(val_ba)
    }

    history.append({
        "Epoch": epoch,
        "Train_Loss": float(train_loss),
        **current_metrics,
        "Validation_Threshold": float(threshold)
    })

    for name, value in current_metrics.items():
        direction = checkpoint_specs[name][1]
        improved = (
            value < best_values[name]
            if direction == "min"
            else value > best_values[name]
        )

        if not improved:
            continue

        best_values[name] = float(value)

        best_info[name] = {
            "epoch": int(epoch),
            "value": float(value),
            "threshold": float(threshold)
        }

        # Python scalar metadata avoids NumPy-scalar checkpoint issues.
        checkpoint_data = {
            "epoch": int(epoch),
            "model_state_dict": copy.deepcopy(model.state_dict()),
            "validation_loss": float(val_loss),
            "validation_roc_auc": float(val_auc),
            "validation_pr_auc": float(val_ap),
            "validation_balanced_accuracy": float(val_ba),
            "validation_threshold": float(threshold)
        }

        torch.save(
            checkpoint_data,
            CHECKPOINT_DIR / checkpoint_specs[name][0]
        )

    if epoch % 50 == 0:
        print(
            f"Epoch {epoch}/{MAX_EPOCHS} | "
            f"Train Loss={train_loss:.4f} | "
            f"Val Loss={val_loss:.4f} | "
            f"ROC-AUC={val_auc:.4f} | "
            f"PR-AUC={val_ap:.4f} | "
            f"BA={val_ba:.4f} | "
            f"Threshold={threshold:.4f}"
        )

history_df = pd.DataFrame(history)
history_df.to_csv(OUT_DIR / "training_history.csv", index=False)

pd.DataFrame([
    {"Checkpoint": name, **info}
    for name, info in best_info.items()
]).to_csv(OUT_DIR / "best_validation_checkpoints.csv", index=False)

# Save training curves.
plt.figure(figsize=(8, 5))
plt.plot(history_df["Epoch"], history_df["Train_Loss"], label="Train")
plt.plot(history_df["Epoch"], history_df["Validation_Loss"], label="Validation")
plt.xlabel("Epoch")
plt.ylabel("Weighted cross-entropy loss")
plt.title("HECKTOR GPT-2 Tabular — Training History")
plt.legend()
plt.tight_layout()
plt.savefig(OUT_DIR / "training_history.png", dpi=150)
plt.show()

# Save fitted preprocessing object.
joblib.dump(preprocessor, OUT_DIR / "tabular_preprocessor.joblib")

# ============================================================
# 11. TEST EVALUATION FOR ALL FOUR CHECKPOINTS
# ============================================================

results = []
predictions_by_model = {}

for name, (filename, _) in checkpoint_specs.items():
    checkpoint_path = CHECKPOINT_DIR / filename

    if not checkpoint_path.exists():
        raise FileNotFoundError(
            f"Checkpoint missing: {checkpoint_path}. "
            "Complete training before running evaluation."
        )

    # Only load checkpoints generated by this trusted notebook.
    checkpoint = torch.load(
        checkpoint_path,
        map_location=device,
        weights_only=False
    )
    model.load_state_dict(checkpoint["model_state_dict"])

    test_scores = predict_probabilities(X_test_t)
    threshold = float(best_info[name]["threshold"])
    test_preds = (test_scores >= threshold).astype(int)

    test_auc = roc_auc_score(y_test, test_scores)
    test_ap = average_precision_score(y_test, test_scores)
    test_ba = balanced_accuracy_score(y_test, test_preds)
    test_f1 = f1_score(y_test, test_preds, zero_division=0)
    cm = confusion_matrix(y_test, test_preds, labels=[0, 1])

    print(
        f"\n=== {name} | Best epoch {best_info[name]['epoch']} "
        f"| Threshold {threshold:.4f} ==="
    )
    print(classification_report(
        y_test,
        test_preds,
        labels=[0, 1],
        target_names=["HPV Negative", "HPV Positive"],
        digits=4,
        zero_division=0
    ))
    print("Confusion matrix [[TN, FP], [FN, TP]]:\n", cm)
    print(
        f"Test ROC-AUC={test_auc:.4f} | "
        f"PR-AUC={test_ap:.4f} | "
        f"BA={test_ba:.4f} | F1={test_f1:.4f}"
    )

    results.append({
        "Dataset": "HECKTOR",
        "Model": "GPT-2 Tabular",
        "Checkpoint_Selected_By": name,
        "Best_Epoch": int(best_info[name]["epoch"]),
        "Validation_Threshold": threshold,
        "Test_ROC_AUC": float(test_auc),
        "Test_PR_AUC": float(test_ap),
        "Test_Balanced_Accuracy": float(test_ba),
        "Test_F1": float(test_f1),
        "TN": int(cm[0, 0]),
        "FP": int(cm[0, 1]),
        "FN": int(cm[1, 0]),
        "TP": int(cm[1, 1])
    })

    predictions_by_model[name] = {
        "scores": test_scores,
        "predictions": test_preds,
        "threshold": threshold
    }

    pd.DataFrame({
        "True_Label": y_test,
        "HPV_Positive_Score": test_scores,
        "Predicted_Label": test_preds,
        "Threshold": threshold
    }).to_csv(
        OUT_DIR / f"test_predictions_{name.lower()}.csv",
        index=False
    )

results_df = pd.DataFrame(results)
results_df.to_csv(OUT_DIR / "test_metrics_by_checkpoint.csv", index=False)

# ============================================================
# 12. ROC AND PRECISION-RECALL CURVES
# ============================================================

for name, pred_data in predictions_by_model.items():
    scores = pred_data["scores"]
    safe_name = name.lower()

    fpr, tpr, roc_thresholds = roc_curve(y_test, scores)
    precision, recall, _ = precision_recall_curve(y_test, scores)

    auc = roc_auc_score(y_test, scores)
    ap = average_precision_score(y_test, scores)

    pd.DataFrame({
        "FPR": fpr,
        "TPR": tpr,
        "Threshold": roc_thresholds
    }).to_csv(OUT_DIR / f"roc_coordinates_{safe_name}.csv", index=False)

    pd.DataFrame({
        "Precision": precision,
        "Recall": recall
    }).to_csv(OUT_DIR / f"pr_coordinates_{safe_name}.csv", index=False)

    plt.figure(figsize=(6, 5))
    plt.plot(fpr, tpr, label=f"AUC={auc:.3f}")
    plt.plot([0, 1], [0, 1], "--")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title(f"HECKTOR ROC — {name}")
    plt.legend()
    plt.tight_layout()
    plt.savefig(OUT_DIR / f"roc_{safe_name}.png", dpi=150)
    plt.show()
    plt.close()

    plt.figure(figsize=(6, 5))
    plt.plot(recall, precision, label=f"AP={ap:.3f}")
    plt.xlabel("Recall")
    plt.ylabel("Precision")
    plt.title(f"HECKTOR Precision-Recall — {name}")
    plt.legend()
    plt.tight_layout()
    plt.savefig(OUT_DIR / f"pr_{safe_name}.png", dpi=150)
    plt.show()
    plt.close()

# ============================================================
# 13. SAVE RUN SUMMARY
# ============================================================

settings = {
    "Dataset": "HECKTOR",
    "Model": MODEL_NAME,
    "Target": TARGET,
    "Features": FEATURES,
    "Numeric_Features": NUMERIC_FEATURES,
    "One_Hot_Encoded_Features": CATEGORICAL_FEATURES,
    "Split": "65/15/20 stratified",
    "Seed": SEED,
    "Epochs": MAX_EPOCHS,
    "Batch_Size": BATCH_SIZE,
    "Learning_Rate": LEARNING_RATE,
    "Weight_Decay": WEIGHT_DECAY,
    "Dropout": DROPOUT,
    "Class_Weighted_Loss": True,
    "SMOTE": False,
    "Hard_Clipping": False,
    "Gradient_Clipping": False,
    "GPT2_Backbone_Frozen": False,
    "Checkpoint_Selection": [
        "Validation loss",
        "Validation ROC-AUC",
        "Validation PR-AUC",
        "Validation balanced accuracy"
    ],
    "Threshold_Selection": (
        "Maximum validation balanced accuracy over unique scores "
        "plus 0 and 1; ties closest to 0.5"
    ),
    "Best_Checkpoints": best_info
}

with open(OUT_DIR / "run_summary.json", "w", encoding="utf-8") as f:
    json.dump(settings, f, indent=2)

# Optional compact comparison CSV, matching Hancock's principal metrics.
comparison = results_df[[
    "Dataset", "Model", "Checkpoint_Selected_By", "Best_Epoch",
    "Test_ROC_AUC", "Test_PR_AUC",
    "Test_Balanced_Accuracy", "Test_F1"
]]
comparison.to_csv(OUT_DIR / "model_comparison.csv", index=False)

print("\nAll outputs saved to:", OUT_DIR.resolve())
print("Result files:", sorted(p.name for p in OUT_DIR.iterdir()))
```
